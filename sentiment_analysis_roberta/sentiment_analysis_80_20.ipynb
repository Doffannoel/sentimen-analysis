```python

import pandas as pd

df = pd.read_csv('../dataset/Data_Gabungan.csv')
df.head()
```

```python

# Sesuaikan nama kolom jika perlu
df = df[['text','label']].dropna()

# Normalisasi label
df['label'] = df['label'].astype(str).str.lower()
df.head()
```

```python

from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df['label'])

print("Train size:", len(train_df))
print("Test size:", len(test_df))
```

```python

from transformers import pipeline

classifier = pipeline(
    "sentiment-analysis",
    model="w11wo/indonesian-roberta-base-sentiment-classifier"
)
```

```python

# Gunakan data test untuk evaluasi
texts = test_df['text'].tolist()
true_labels = test_df['label'].tolist()

predictions = classifier(texts, truncation=True)

pred_labels = [p['label'].lower() for p in predictions]
```

```python

from sklearn.metrics import confusion_matrix, precision_score, recall_score

labels = ['positive','negative']
cm = confusion_matrix(true_labels, pred_labels, labels=labels)

TP = cm[0][0]
FN = cm[0][1]
FP = cm[1][0]
TN = cm[1][1]

precision = precision_score(true_labels, pred_labels, pos_label='positive')
recall = recall_score(true_labels, pred_labels, pos_label='positive')

print("Confusion Matrix:\n", cm)
print("\nTrue Positive (TP):", TP)
print("True Negative (TN):", TN)
print("False Positive (FP):", FP)
print("False Negative (FN):", FN)

print("\nPrecision:", precision)
print("Recall:", recall)
```
